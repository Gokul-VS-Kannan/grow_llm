```python
import ollama
```

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role":"system",
            "content":"You are a helpful teacher who explains things simply. Use short sentence. Giveone example for every cocept",
        },
        {
            "role":"user",
            "content":"What is variable in programming? /no_think"
        }
    ]
)

print(response["message"]["content"])
```

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "system", 
            "content": "You are a friendly math tutor for children."
        },
        {
            "role": "user", 
            "content": "What is 5 + 3? /no_think"
        },
        {
            "role": "assistant", 
            "content": "5 + 3 = 8. You have 5 apples, someone gives 3 more. Now you have 8!"
        },
        {
            "role": "user", 
            "content": "Now multiply that answer by 2. /no_think"
        }
    ]
)
print(response["message"]["content"])
```

*Zero Shot Prompting*

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user",
            "content": "Classify the following text as positive, negative, or neutral: 'I love this phone, the camera is amazing!' /no_think"
        }
    ]
)
print(response["message"]["content"])
```

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user",
            "content": "Extract the product name, price and store from this text:'The Samsung Galaxy S24 is available for RS 74,999 on Flipkart.'Return it as a clean list /no_think"
        }
    ]
)
print(response["message"]["content"])
```
