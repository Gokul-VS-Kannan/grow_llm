*Few-Shot Text Classification*

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "system",
            "content": "You are a text classifier. Classify text as Positive, Negative, or Neutral. Reply with ONLY the label, nothing else."
        },
        {
            "role": "user", 
            "content": "The food was delicious and the service was excellent!"
        },
        {
            "role": "assistant", 
            "content": "Positive"
        },
        {
            "role": "user", 
            "content": "The delivery was late and the item was broken."
        },
        {
            "role": "assistant", 
            "content": "Negative"
        },
        {
            "role": "user", 
            "content": "I received the package today."
        },
        {
            "role": "assistant", 
            "content": "Neutral"
        },
        {
            "role": "user", 
            "content": "This movie is absolutely wonderful, I cried tears of joy! /no_think"
        }
    ]
)
print(response["message"]["content"])
```

*Chain Of Thought*

```python
# Direct (no reasoning shown)
response1 = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user", 
            "content": "A fruit seller has 3 baskets. Each basket has 15 mangoes. He sells 7 in the morning and 12 in the evening. How many are left? Just give the answer. /no_think"
        }
    ]
)
print("Direct:", response1["message"]["content"])
```

```python
# Chain-of-thought (reasoning shown)
response2 = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user", 
            "content": "A fruit seller has 3 baskets. Each basket has 15 mangoes. He sells 7 in the morning and 12 in the evening. How many are left? Think step by step. /no_think"
        }
    ]
)
print("CoT:", response2["message"]["content"])
```

*COT and few shot*

```python
response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "system", 
            "content": "You solve word problems step by step. Show each step, then the answer."
        },
        {
            "role": "user", 
            "content": "A teacher has 4 packets of pens. Each has 10. She gives 6 away. How many left?"
        },
        {
            "role": "assistant", 
            "content": "Step 1: Total pens = 4 x 10 = 40\nStep 2: Given away = 6\nStep 3: Remaining = 40 - 6 = 34\n\nAnswer: 34 pens"
        },
        {
            "role": "user", 
            "content": "A shop has 5 shelves with 8 books each. It receives 15 new books but sells 20. How many now? /no_think"
        }
    ]
)
print(response["message"]["content"])
```

```python
import json
```

```python
response = ollama.chat(
    model = "qwen3:0.6b",
    messages=[
         {
            "role": "system",
            "content": "Extract information and return it as JSON with keys: name, age, city."
        },
        {
            'role':'user',
            'content':"Rahul is 25 years old and lives in Bangalore./no_think"
        }
    ],

    format='json'
)

raw = response['message']['content']
print('Raw response:')
print(raw)

data = json.loads(raw)

print(f"Name: {data['name']}")
print(f"Age: {data['age']}")
print(f"City: {data['city']}")
```
