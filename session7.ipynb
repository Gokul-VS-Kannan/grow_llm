```python

response = ollama.chat(
    model = "qwen3:0.6b",
    messages = [
        {
            "role":"user",
            "content":"what is machine learning? Answer in 2 sentence."
        }
    ]
)

answer = response["message"]["content"]

print('Answer:')
print(answer)
print()

# calculating speed
# duration in nano seconds

total_seconds = response['total_duration']/1_000_000_000
eval_seconds = response['eval_duration']/1_000_000_000
tokens_generated = response['eval_count']
tokens_per_second = tokens_generated/eval_seconds

print(f"Total time: {total_seconds:.2f} seconds")
print(f"Tokens generated: {tokens_generated}")
print(f"Generation speed: {tokens_per_second:.1f} tokens/second")
```

```python
print("Question : What are the 3 most popular programming language?")
print()
stream = ollama.chat(
    model = "qwen3:0.6b",
    messages = [
        {
            'role':'user',
            'content':'What are the 3 most popular programming language'
        }
    ],
    stream = True
)

for chunk in stream:
    print(chunk['message']['content'], end="",flush=True)
print()
```
