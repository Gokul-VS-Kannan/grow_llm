```python

response = ollama.chat(
    model = "qwen3:0.6b",
    messages = [
        {
            "role":"user",
            "content":"what is machine learning? Answer in 2 sentence."
        }
    ]
)

answer = response["message"]["content"]

print('Answer:')
print(answer)
print()

# calculating speed
# duration in nano seconds

total_seconds = response['total_duration']/1_000_000_000
eval_seconds = response['eval_duration']/1_000_000_000
tokens_generated = response['eval_count']
tokens_per_second = tokens_generated/eval_seconds

print(f"Total time: {total_seconds:.2f} seconds")
print(f"Tokens generated: {tokens_generated}")
print(f"Generation speed: {tokens_per_second:.1f} tokens/second")
```

```python
print("Question : What are the 3 most popular programming language?")
print()
stream = ollama.chat(
    model = "qwen3:0.6b",
    messages = [
        {
            'role':'user',
            'content':'What are the 3 most popular programming language'
        }
    ],
    stream = True
)

for chunk in stream:
    print(chunk['message']['content'], end="",flush=True)
print()
```

```python
conversation = [
    {"role": "user", "content": "My name is Ravi."},
]

# First message
response1 = ollama.chat(
    model="qwen3:0.6b",
    messages=conversation
)

reply1 = response1.message.content

# Add the model's reply to the history
conversation.append({
    "role": "assistant",
    "content": reply1
})

# Add the next user message
conversation.append({
    "role": "user",
    "content": "What is my name?"
})

# Second message -- send FULL history
response2 = ollama.chat(
    model="qwen3:0.6b",
    messages=conversation
)

print(response2['message']['content'])
```
