*Simple Neural Network with 3 Neurons*

```python
import torch.nn as nn

class SimpleNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # layer 1 take 10 input produce 5 out
        self.layer1 = nn.Linear(10,5)
        # layer 2 take 5 input produce 3 out
        self.layer2 = nn.Linear(5,3)
        # layer 3 take 3 input produce 1 out
        self.layer3 = nn.Linear(3,1)

    def forward(self,x):
        output1 = self.layer1(x)
        output2 = self.layer2(output1)
        output3 = self.layer3(output2)
        return output3

model = SimpleNetwork()
print(model)
```

```python
total_parameters = sum(p.numel() for p in model.parameters()) # numel number of elements

print(total_parameters)
```
