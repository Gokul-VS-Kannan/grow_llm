*Simple Neural Network with 3 Neurons*

```python
import torch.nn as nn

class SimpleNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # layer 1 take 10 input produce 5 out
        self.layer1 = nn.Linear(10,5)
        # layer 2 take 5 input produce 3 out
        self.layer2 = nn.Linear(5,3)
        # layer 3 take 3 input produce 1 out
        self.layer3 = nn.Linear(3,1)

    def forward(self,x):
        output1 = self.layer1(x)
        output2 = self.layer2(output1)
        output3 = self.layer3(output2)
        return output3

model = SimpleNetwork()
print(model)
```

```python
total_parameters = sum(p.numel() for p in model.parameters()) # numel number of elements

print(total_parameters)
```

```python
from transformers import  AutoTokenizer

# load tokanizer from a well known model

tokenizer = AutoTokenizer.from_pretrained("gpt2")
text = 'Hello I am learning about AI and LLMs'
tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print(f"Original text : '{text}' ")
print(f"Tokens : {tokens}")
print(f"Number of tokens : {len(tokens)}")
print(f"Token IDs : {token_ids}")
```

*Embedding*

```python
import ollama
import numpy as np

texts = [
    "I love programming in python",
    "Python is so simple and beginner friendly programming language",
    "The weather is so hot today"
]

embeddings = []

for text in texts:

    result = ollama.embed(
        model="nomic-embed-text",
        input=text
    )

    embeddings.append(result["embeddings"][0])


embeddings = [np.array(e) for e in embeddings]

print(f"Each embedding has {len(embeddings[0])} numbers")
print(f"First 10 numbers of sentence 1: {embeddings[0][:10]}")
print()


def cosine_similarity(a, b):

    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )


sim_1_2 = cosine_similarity(
    embeddings[0],
    embeddings[1]
)

sim_1_3 = cosine_similarity(
    embeddings[0],
    embeddings[2]
)


print(f"Python vs Python: {sim_1_2:.4f}")
print(f"Python vs Weather: {sim_1_3:.4f}")

print()
print("Higher number = more similar meaning")
```
