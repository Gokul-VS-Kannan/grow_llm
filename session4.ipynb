```python
import time

def ask_ai(question, model ="qwen3:0.6b"):
    response = ollama.chat(
        model = model,
        messages = [
            {
                "role" : "user",
                "content" : question
            }
        ]
    )

    return response["message"]["content"]

start = time.time()
answer = ask_ai("What is Python?")
end = time.time()

print(f"Answer : {answer}")
print(f"Time taken : {end - start:.2f} seconds")
```

```python
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(f"[TIMER] {func.__name__} took {end - start:.2f} seconds")
        return result
    return wrapper
```

```python
@timer
def ask_ai_timed(question, model ="qwen3:0.6b"):
    response = ollama.chat(
        model = model,
        messages = [
            {
                "role" : "user",
                "content" : question
            }
        ]
    )

    return response["message"]["content"]
answer = ask_ai_timed("What is List in python? one sentence")
print(f"Answer : {answer}")
```

```python
def retry(max_attempt =3):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for attempt in range(1,max_attempt+1):
                try:
                    func(*args,**kwargs)
                except Exception as e:
                    print(f"[RETRY] Attempt {attempt}")

                    if attempt==max_attempt:
                        print(f"[RETRY] All {max_attempt} attempt failed")
                        raise
 
        return wrapper
    return decorator

@retry(max_attempt=3)
def risky_ai_call(question):
    import random

    if random.random() < 0.9:
        raise ConnectionError("Model busy try again")
    return ask_ai(question)

try :
    answer = risky_ai_call("What is the capital of Tndia?")
    print(f"Success: {answer}")

except ConnectionError:
    print("No more Rety Token expired")
```

```python
import asyncio 


async def ask_ai_async(question, model = 'qwen3:0.6b'):
    """ ask_ai model in asynchronus way"""
    messages = [
        {
            'role' : 'system',
            'content' : 'You are trainer who expain content very easy'
        },
        {
            'role' : 'user',
            'content' : question,
        },
    ]
    response = ollama.chat(
        model = model,
        messages = messages,
    )

    return response['message']['content']

# ask question in synchronus way

question = [
    'What is python? One sentence',
    'What is JSON? One sentence',
    'What is API? One sentence,'
]

print('===== One at a time (sync) =====')
start = time.time()
for q in question:
    answer = ask_ai(q)
    print(f'Q: {q}')
    print(f'A:{answer[:80]}')
    print()
end = time.time()
print(f'Sync total : {end - start:.2f} seconds')
```
