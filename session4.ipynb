```python
prompt_text = "You are an helpfull AI trainer who explain concepts in a simple way"\

with open("sample.txt","w") as file:
    
    file.write(prompt_text)

print("file created: sample.txt")
```

```python
with open("sample.txt", "r") as file:
    content = file.read()

print("file content")
print(content)
```

```python
import ollama
with open("sample.txt", "r") as file:
    system_prompt = file.read()

response = ollama.chat(
    model = "qwen3:0.6b",
    messages =[
        {
            "role" : "system",
            "content" : system_prompt,
        },
        {
            "role" : "user",
            "content" : "what is a neural network"
        },
    ]
)

print(response["message"]["content"])
```

```python
answer = response["message"]["content"]

with open("ai_response.txt", "w", encoding="utf-8") as file:
    file.write(answer)

print("file createsd : ai_response.txt")

with open("ai_response.txt","r", encoding="utf-8") as file:
    source = file.read()

print("file content is")
print(source)
```

```python
import json

model_config = {
    "model_name" : "qwen3:0.6B",
    "temperature" : 0.7,
    "max_token" : 500,
    "system_prompt" : "You are a helpful asistant",
    "available" : True,
}

json_string = json.dumps(model_config, indent=2)
print(json_string)
```

```python
with open("model_config.json","w") as file:
    json.dump(model_config, file, indent=2)
print("Saved to model_config.json")
```

```python
with open("model_config.json", "r") as file:
    load_config = json.load(file)

print(load_config)
print(type(load_config))
```

```python
import yaml

config={
    "project" : "AI Tutor Bot",
    "model" : {
        "model_name" : "qwen3:0.6B",
        "temperature" : 0.7,
        "max_token" : 500,
    },
    "features" : ["chat", "summarize", "translate"],
}

with open("config.yaml", "w") as file:
    yaml.dump(config, file, default_flow_style=False)

print("Saved config.yaml")
```

```python
with open("config.yaml", "r") as file:
    load_yaml = yaml.safe_load(file)

print(load_yaml)
```
